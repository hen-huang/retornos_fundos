# Prévia de cotas por fatores

Estima o retorno de hoje dos fundos Long Only, Long Biased, Long Short e Macro com PL ≥ R$ 100 mi,
antes da divulgação da cota. Cada fundo é estimado em seis janelas (1m, 2m, 3m, 6m, 12m, 24m) e num modelo exponencial (EW), e cada prévia
traz um intervalo de 95% bicaudal.

Rode as células em ordem com Shift + Enter. Atualizar e estimar: 1 vez por dia, depois das 8h.

## Configuração

```python
import previa as pv

ATUALIZAR_HOJE = True

pv.PL_MINIMO = 100e6
pv.JANELA_PRINCIPAL = "comb"     # combinada; ou "1m", "2m", "3m", "6m", "12m", "24m", "ew"
pv.OBS_MIN_ADICIONAIS = 120      # fatores adicionais só nos modelos com >= 120 obs.; 0 = em todos
pv.DATAS_EXCLUIR = ["2026-10-26"]
```

## 1. Atualizar a base

```python
if ATUALIZAR_HOJE:
    pv.atualizar()
```

## 2. Estimar
Estima os βs, calibra o IC e calcula os pesos da combinada pelo erro fora da amostra.

```python
pv.estimar()
```

## 3. Backtest
Cada dia é previsto só com dados até a véspera. Erro em bps (1 bp = 0,01%); cobertura ideal ≈ 95%.

```python
pv.backtest()
```
